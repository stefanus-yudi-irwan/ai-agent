```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display

load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print("OPENAI_API_KEY exists")
else:
    print("OPENAI_API_KEY didn't exists")

OPENAI_MODEL = "gpt-4.1-nano"
openai = OpenAI()
```

```python
request = "Please come up with challenging, nuanced question that I can ask a number of LLMs to evaluate their intelligence"
request += "Answer only with the question, no explanation."
messages = [{"role": "user", "content": request}]
messages
```

```python
response = openai.chat.completions.create(
    model = OPENAI_MODEL,
    messages = messages
)
question = response.choices[0].message.content
print(question)
```

```python
competitors = []
answers = []

messages = [{
    "role": "user",
    "content": question,
}]

response = openai.chat.completions.create(
    model = OPENAI_MODEL,
    messages = messages
)

answer = response.choices[0].message.content
display(Markdown(answer))
competitors.append(OPENAI_MODEL)
answers.append(answer)
```
