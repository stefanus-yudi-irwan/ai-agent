```python
from dotenv import load_dotenv
from openai import OpenAI
from pypdf import PdfReader
import gradio as gr
import os
from IPython.display import Markdown, display

load_dotenv(override=True)
openai = OpenAI(
    api_key = os.getenv("OPENAI_API_KEY")
)
NAME = "Stefanus Yudi Irwan"
OPENAI_MODEL = "gpt-4.1-nano"
```

```python
reader = PdfReader("data/Resume - Stefanus Yudi Irwan.pdf")
resume = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        resume += text
print(resume)
```

```python
system_prompt = f"You are acting as {NAME}. You are answering questions on {NAME}'s website, \
    particularly questions related to {NAME}'s career, background, skills, and experience. \
    Your responsibility is to represent {NAME} for interactions on the website as faithfully as possible. \
    Your are given a summary of {NAME}'s background and resume profile which you can use to answer questions. \
    Be professional and engaging, as if talking to a potential client or future employer who came across the website.\
    If you don't know the answer, say so."

system_prompt += f"\n\n##Resume: \n\n{resume}\n\n"
system_prompt += f"With this context, please chat with the user, always staying in character as {NAME}"

display(Markdown(system_prompt))
```

```python
def chat(message: str, history: str) -> str:
    """Fallback function for gradio chat interface
    Input:
    message (str): message input from user
    history (str): history message from AI model
    Return:
    (str): response from AI model
    """
    input_message = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    model_response = openai.chat.completions.create(model=OPENAI_MODEL, messages=input_message)
    output_message = model_response.choices[0].message.content
    return output_message
```

```python
gr.ChatInterface(chat, type="messages").launch()
```

```python
from pydantic import BaseModel

class Evaluation(BaseModel):
    is_acceptable: bool
    feedback: str
    
evaluator_system_prompt = f"You are an evaluator that decides whether a response to a question is acceptable. \
    You are provided with a conversation between a User and an Agent. Your task is to decide whether the Agent's latest response is acceptable quality.\
    The Agent is playing the role of {NAME} and is representing {NAME} on their website.\
    The Agent has been instructed to be professional and engaging, as if talking to a potential client or future employer who came across the website.\
    The Agent has been provided with context on {NAME} in the form of their resume details. Here's the information:"

evaluator_system_prompt += f"\n\n##Resume: \n\n{resume}\n\n"
evaluator_system_prompt += f"With this context, please evaluate the latest response, replying with whether the response is acceptable and your feedback"
display(Markdown(evaluator_system_prompt))
```

```python
def evaluator_user_prompt(reply: str, message: str, history: str) -> str:
    """
    Function to create user prompt for evaluator 
    Input 
    reply (str): message from AI regarding user questions 
    message (str): latest message from AI
    history (str): history of chat between user and AI 
    Output 
    (str): prompt for evaluator user
    """
    user_prompt = f"Here's the conversation between the User and the Agent: \n\n{history}\n\n"
    user_prompt += f"Here's the latest message from the User: \n\n{message}\n\n"
    user_prompt += f"Here's the latest response from the Agent:\n\n{reply}\n\n"
    user_prompt += "Please evaluate the response, replying with whether it is acceptable and your feedback."
    return user_prompt

def evaluate(reply: str, message: str, history: str) -> str:
    """Fallback function for evaluator
    Input 
    reply (str): message from AI regarding user questions
    message (str): message question from the user
    history (str): all the history chat from the AI 
    Output 
    (str): response from the AI model
    """
    input_message = [{"role": "system", "content": evaluator_system_prompt}] + [{"role": "user", "content": evaluator_user_prompt(reply, message, history)}]
    model_response = openai.beta.chat.completions.parse(model=OPENAI_MODEL, messages=input_message, response_format=Evaluation)
    output_message: Evaluation = model_response.choices[0].message.parsed
    return output_message

def rerun(reply: str, message: str, history: str, feedback: str) -> str:
    """Function to re run the unpassed response to AI model 
    Input 
    reply (str): message from AI regarding user questions
    message (str): current message from AI model 
    history (str): all the history chat from the AI 
    feedback (str): feedback from the AI model
    """
    updated_system_prompt = system_prompt + "\n\n## Previous answer rejected\nYou just tried to reply, but the quality control reject your reply\n"
    updated_system_prompt += f"## Your attempted answer:\n{reply}\n\n"
    updated_system_prompt += f"## Reason for rejection:\n{feedback}\n\n"
    messages = [{"role": "system", "content": updated_system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=OPENAI_MODEL, messages=messages)
    response_text = response.choices[0].message.content
    return response_text

def chat(message: str, history: str) -> str:
    """Fallback function for chat gradio
    Input
    message (str): message input for AI model
    history (str): history chat in AI model
    Output
    (str): reponse from AI model
    """
    if "patent" in message:
        system = system_prompt + "\n\nEverything in your reply needs to be in pig latin -\
            it is mandatory that your respond only and entirely in pig latin"
    else:
        system = system_prompt

    messages = [{"role": "system", "content": system}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=OPENAI_MODEL, messages=messages)
    reply = response.choices[0].message.content

    evaluation = evaluate(reply, message, history)

    if evaluation.is_acceptable:
        print("Passed evaluation - returning reply")
    else:
        print("Failed evaluation - retrying")
        print(evaluation.feedback)
        reply = rerun(reply, message, history, evaluation.feedback)
    return reply
```

```python
message_test_evaluate = [{"role": "system", "content": system_prompt},
    {"role": "user", "content": "do you hold a patent?"}]
response = openai.chat.completions.create(model=OPENAI_MODEL, messages=message_test_evaluate)
reply_text = response.choices[0].message.content
print(f"AI Model reply: {reply_text}")
evaluation_result = evaluate(reply_text, "do you hold a patent?", message_test_evaluate[:1])
print("=============================")
evaluation_result
```
